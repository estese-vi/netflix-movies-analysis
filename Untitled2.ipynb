```python
from sklearn.datasets import load_iris
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats


iris = load_iris()
data = iris.data
petal_length = iris.data[:, 2]  # все строки, третий столбец (индекс 2)

# Нормированная гистограмма
plt.hist(petal_length, bins=30, density=True,
         alpha=0.7, color='skyblue', edgecolor='black',
         label='Гистограмма')

# График плотности (KDE)
kde = stats.gaussian_kde(petal_length)
x_vals = np.linspace(petal_length.min(), petal_length.max(), 200)
plt.fill_between(x_vals, kde(x_vals), color='red', alpha=0.1)
plt.plot(x_vals, kde(x_vals), color='red', linewidth=2, label='График плотности (KDE)')

plt.xlabel('Длина лепестка (см)')
plt.ylabel('Плотность вероятности')
plt.title('Распределение длины лепестка (нормированная гистограмма + KDE)')
plt.legend()
plt.show()

```

```python
from sklearn.datasets import load_iris
import matplotlib.pyplot as plt
import numpy as np


iris = load_iris()
data = iris.data
target = iris.target

# Разделение данных по классам
setosa = data[target == 0, 2]        # длина лепестка для setosa
versicolor = data[target == 1, 2]    # длина лепестка для versicolor
virginica = data[target == 2, 2]     # длина лепестка для virginica


plt.hist(setosa, bins=15, edgecolor='black', color='red', label='Setosa')
plt.hist(versicolor, bins=15, edgecolor='black', color='green', label='Versicolor')
plt.hist(virginica, bins=15, edgecolor='black', color='purple', label='Virginica')
plt.title('Гистограмма распределения длины лепестков для 3 видов')
plt.xlabel('Длина лепестка')
plt.ylabel('Количество')
plt.grid(axis='y', alpha=0.3)
plt.legend(loc='upper right')
plt.show()


# График 2: Графики плотности для трёх видов
# KDE для Setosa
kde_setosa = stats.gaussian_kde(setosa)
x_setosa = np.linspace(setosa.min() - 0.5, setosa.max() + 0.5, 200)
plt.plot(x_setosa, kde_setosa(x_setosa), color='red', linewidth=2, label='Setosa')
plt.fill_between(x_setosa, kde_setosa(x_setosa), color='red', alpha=0.1)

# KDE для Versicolor
kde_versicolor = stats.gaussian_kde(versicolor)
x_versicolor = np.linspace(versicolor.min() - 0.5, versicolor.max() + 0.5, 200)
plt.plot(x_versicolor, kde_versicolor(x_versicolor), color='green', linewidth=2, label='Versicolor')
plt.fill_between(x_versicolor, kde_versicolor(x_versicolor), color='green', alpha=0.1)

# KDE для Virginica
kde_virginica = stats.gaussian_kde(virginica)
x_virginica = np.linspace(virginica.min() - 0.5, virginica.max() + 0.5, 200)
plt.plot(x_virginica, kde_virginica(x_virginica), color='blue', linewidth=2, label='Virginica')
plt.fill_between(x_virginica, kde_virginica(x_virginica), color='blue', alpha=0.1)
plt.legend()
plt.xlabel('Длина лепестка')
plt.ylabel('Плотность вероятности')
plt.title('Графики плотности для трёх видов')
plt.show()
```

```python
import numpy as np
from sklearn.datasets import load_diabetes
import matplotlib.pyplot as plt

# Загрузка данных
diabetes = load_diabetes()
data = diabetes.data
age = diabetes.data[:, 0]  # признак age
target = diabetes.target    # прогрессия диабета

median_age = np.median(age)
young_patients = target[age <= median_age]
old_patients = target[age > median_age]

# Построение двух ящиков на одном графике
# Создаем список данных для boxplot
data_to_plot = [young_patients, old_patients]

# Настройка цветов для каждого ящика
box = plt.boxplot(data_to_plot,
                  patch_artist=True,
                  notch=True,  # выемка у медианы
                  labels=['Молодые пациенты', 'Пожилые пациенты'])

# Установка цветов заливки для каждого ящика
box['boxes'][0].set_facecolor('lightgreen')
box['boxes'][1].set_facecolor('lightskyblue')

plt.ylabel('Прогрессия диабета')
plt.title('Сравнение прогрессии диабета в разных возрастных группах')
plt.show()
```

```python
from sklearn.datasets import load_breast_cancer
import matplotlib.pyplot as plt
import numpy as np

# Загрузка данных
cancer = load_breast_cancer()
data = cancer.data

# Извлекаем нужные столбцы
mean_radius = data[:, 0]       # средний радиус
mean_area = data[:, 3]         # средняя площадь
mean_smoothness = data[:, 4]   # средняя гладкость
mean_concavity = data[:, 6]    # средняя вогнутость

plt.subplots(figsize=(14, 7))
plt.scatter(mean_radius, mean_area, alpha=0.6, s=60, c='blue', edgecolor='black')
x_smooth = np.linspace(mean_radius.min(), mean_radius.max(), 100)

#  Рисуем линию тренда
coefficients = np.polyfit(mean_radius, mean_area, 1)  # 1 — линейная зависимость
#  Создаём функцию для расчёта значений на линии тренда
trend_line = np.poly1d(coefficients)
plt.plot(x_smooth, trend_line(x_smooth),
         color='red', linewidth=2.5, linestyle='--')

plt.title('Зависимость площади опухоли от радиуса', fontsize=14, fontweight='bold')
plt.xlabel('Средний радиус', fontsize=12)
plt.ylabel('Средняя площадь', fontsize=12)
plt.grid(True, alpha=0.3)
plt.show()
```

```python
from sklearn.datasets import load_breast_cancer
import matplotlib.pyplot as plt
import numpy as np

# Загрузка данных
cancer = load_breast_cancer()
data = cancer.data
target = cancer.target

mean_radius = data[:, 0]  # средний радиус
mean_area = data[:, 3]    # средняя площадь

# Создание отдельных массивов для злокачественных и доброкачественных опухолей
radius_malignant = mean_radius[target == 0]
area_malignant = mean_area[target == 0]
radius_benign = mean_radius[target == 1]
area_benign = mean_area[target == 1]

plt.subplots(figsize=(14, 7))
plt.scatter(radius_malignant, area_malignant,
            c='red', alpha=0.6, s=70,
            marker='o', label='malignant')
plt.scatter(radius_benign, area_benign,
            c='blue', alpha=0.6, s=70,
            marker='s', label='benign')
plt.title('Сравнение размеров опухолей по типам', fontsize=14, fontweight='bold')
plt.xlabel('Средний радиус', fontsize=12)
plt.ylabel('Средняя площадь', fontsize=12)
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

```python
from sklearn.datasets import load_breast_cancer
import matplotlib.pyplot as plt
import numpy as np

# Загрузка данных
cancer = load_breast_cancer()
data = cancer.data

# Извлекаем нужные столбцы
mean_radius = data[:, 0]       # средний радиус
mean_area = data[:, 3]         # средняя площадь
mean_smoothness = data[:, 4]   # средняя гладкость
mean_concavity = data[:, 6]    # средняя вогнутость

plt.subplots(figsize=(14, 7))
scatter = plt.scatter(mean_radius, mean_area,
                      c=mean_smoothness,           # цвет зависит от возраста
                      alpha=0.6,
                      s=80,
                      edgecolor='black')

plt.colorbar(scatter, label='Средняя гладкость')  # добавляем цветовую шкалу

plt.title('Зависимость четырёх признаков опухолей', fontsize=14)
plt.xlabel('Средний радиус', fontsize=12)
plt.ylabel('Средняя площадь', fontsize=12)
plt.grid(True, alpha=0.3)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

# Загрузка данных
iris = load_iris()
data = iris.data
target = iris.target
target_names = iris.target_names
feature_names = iris.feature_names

# Извлечение признаков
sepal_length = data[:, 0]  # длина чашелистика
sepal_width = data[:, 1]   # ширина чашелистика
petal_length = data[:, 2]  # длина лепестка
petal_width = data[:, 3]   # ширина лепестка
unique, counts = np.unique(target, return_counts=True)
sepal_length_setosa = sepal_length[target == 0]      # вид setosa (индекс 0)
sepal_length_versicolor = sepal_length[target == 1]  # вид versicolor (индекс 1)
sepal_length_virginica = sepal_length[target == 2]   # вид virginica (индекс 2)
petal_length_setosa = petal_length[target == 0]
petal_width_setosa = petal_width[target == 0]

petal_length_versicolor = petal_length[target == 1]
petal_width_versicolor = petal_width[target == 1]

petal_length_virginica = petal_length[target == 2]
petal_width_virginica = petal_width[target == 2]

fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(10, 8))

# Строим график в левом верхнем углу
axes[0, 0].pie(counts,
        labels=target_names,
        autopct='%1.1f%%')
axes[0, 0].set_title('Распределение видов ирисов')

# Правый верхний
axes[0, 1].hist(sepal_length, bins=25, color='purple', alpha=0.7, edgecolor='white')
axes[0, 1].set_title('Распределение длины чашелистика у ирисов')
axes[0, 1].set_xlabel('Длина чашелистика (см)')
axes[0, 1].set_ylabel('Частота')
axes[0, 1].grid(True, alpha=0.3)

# Левый нижний
data_to_plot = [sepal_length_setosa, sepal_length_versicolor, sepal_length_virginica]
box = axes[1, 0].boxplot(data_to_plot,
                         patch_artist=True,
                         labels=target_names)

# Установка цветов заливки
colors_box = ['red', 'green', 'blue']
for patch, color in zip(box['boxes'], colors_box):
    patch.set_facecolor(color)

axes[1, 0].set_title('Сравнение длины чашелистика по видам')
axes[1, 0].grid(axis='x', linestyle='--', alpha=0.3)


# Правый нижний
axes[1, 1].scatter(petal_width_setosa, petal_length_setosa,
            color='red', label=target_names[0])

axes[1, 1].scatter(petal_width_versicolor, petal_length_versicolor,
            color='green', label=target_names[1])

axes[1, 1].scatter(petal_width_virginica, petal_length_virginica,
            color='blue', label=target_names[2])

axes[1, 1].set_title('Зависимость длины лепестка от ширины')
axes[1, 1].set_xlabel('Ширина лепестка (см)')
axes[1, 1].set_ylabel('Длина лепестка (см)')
axes[1, 1].grid(True, alpha=0.3)


plt.suptitle('Комплексный анализ датасета Iris')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt


# Loading data
url = "https://raw.githubusercontent.com/Pesimistixxx/NetflixMoviesDataset/master/netflix_titles.csv"
df = pd.read_csv(url, engine='python', encoding='utf-8')

# If automate downloading doesn't work
# Link for dataset downloading: https://www.kaggle.com/datasets/nalisha/netflix-movies-and-tv-shows-data-analysis/data
# df = pd.read_csv('mymoviedb (1).csv')

numeric_columns = ['Popularity', 'Vote_Count', 'Vote_Average']
for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors='coerce')
df = df.dropna(subset=['Vote_Count', 'Vote_Average'])

df['Release_Date'] = pd.to_datetime(df['Release_Date'], errors='coerce')
df['Year'] = df['Release_Date'].dt.year
df['Month'] = df['Release_Date'].dt.month
df['Day'] = df['Release_Date'].dt.day
df['Weekday'] = df['Release_Date'].dt.day_name()

def popularity_category(count):
    if pd.isna(count):
        return 'Unknown'
    elif count < 100:
        return 'Low (<100)'
    elif count < 1000:
        return 'Medium (100-999)'
    elif count < 5000:
        return 'High (1000-4999)'
    else:
        return 'Very High (5000+)'

df['Popularity_Category'] = df['Vote_Count'].apply(popularity_category)
df = df[df['Vote_Count'] > 0]

df['Popularity_Level'] = pd.cut(df['Popularity'],
                                 bins=[0, 100, 500, 1000, df['Popularity'].max()],
                                 labels=['Low', 'Medium', 'High', 'Very High'])

df['Genre_List'] = df['Genre'].str.split(', ')
df_exploded = df.explode('Genre_List')
df['Genre_Count'] = df['Genre'].str.count(',') + 1
df['Main_Genre'] = df['Genre'].str.split(', ').str[0]

lang_counts = df['Original_Language'].value_counts()
top5 = lang_counts.head(5)
# Sum the numbers of others languages (starting from 5-th place)
other_count = lang_counts.iloc[5:].sum()
# Create an element for "Other"
other_series = pd.Series([other_count], index=['Other'])
# Concatenate top-5 and "Other" in one common structure
lang_counts_with_other = pd.concat([top5, other_series])


rare_langs = lang_counts[lang_counts < 10].index
df['Language_Group'] = df['Original_Language'].apply(
    lambda x: 'Other' if x in rare_langs else str(x).upper()
)
df['Title'] = df['Title'].str.strip()
df['Overview'] = df['Overview'].str.strip()
df['Success_Score'] = (df['Popularity'] / df['Popularity'].max() * 50 +
                        df['Vote_Average'] / 10 * 50)
df['Title_Length'] = df['Title'].str.len()
df['Year_Month'] = df['Release_Date'].dt.to_period('M')
df['Quarter'] = df['Release_Date'].dt.quarter
df_yearly_rating = df.groupby('Year')['Success_Score'].mean().reset_index()
df_genre_rating = df.groupby('Main_Genre')['Success_Score'].mean().reset_index()
df_genre_rating = df_genre_rating.sort_values(by='Success_Score', ascending=False)

fig, axes = plt.subplots(nrows=3, ncols=2, figsize=(12, 14))

# 1. Language distribution
axes[0, 0].pie(
    lang_counts_with_other,
    labels=lang_counts_with_other.index,
    autopct='%1.1f%%',
    pctdistance=0.8,
    startangle=140,
    wedgeprops={'edgecolor': 'white', 'linewidth': 1}
)
axes[0, 0].set_title('Language Distribution (Top 5)')

# 2. Success Score per Year
axes[0, 1].plot(df_yearly_rating['Year'], df_yearly_rating['Success_Score'], color='#2b5c8f')
axes[0, 1].set_title('Average Success Score by Year')
axes[0, 1].set_xlabel('Year')
axes[0, 1].set_ylabel('Success Score')
axes[0, 1].grid(True, alpha=0.3)

# 3. Success Score per Genre
axes[1, 0].bar(df_genre_rating['Main_Genre'], df_genre_rating['Success_Score'], color='#f49ac2')
axes[1, 0].tick_params(axis='x', rotation=45)
for label in axes[1, 0].get_xticklabels():
    label.set_ha('right')
axes[1, 0].set_title('Average Success Score by Genre')
axes[1, 0].set_xlabel('Genre')
axes[1, 0].set_ylabel('Success Score')
axes[1, 0].grid(axis='y', alpha=0.3)

# 4. Vote Average vs. Popularity
axes[1, 1].scatter(df['Popularity'], df['Vote_Average'], color='royalblue', alpha=0.4, s=15)
axes[1, 1].set_xscale('log')
axes[1, 1].set_xlabel('Popularity (Log Scale)')
axes[1, 1].set_ylabel('Vote Average')
axes[1, 1].set_title('Vote Average vs. Popularity')
axes[1, 1].grid(True, alpha=0.3)

# 5. Releases by Day of the Week
weekday_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
weekday_counts = df['Weekday'].value_counts().reindex(weekday_order)

axes[2,0].bar(weekday_counts.index, weekday_counts.values, color='#4c72b0')
axes[2,0].set_title('Movie Releases by Day of the Week')
axes[2,0].set_xlabel('Weekday')
axes[2,0].set_ylabel('Number of Movies')
axes[2,0].tick_params(axis='x', rotation=30)
axes[2,0].grid(axis='y', alpha=0.3)

# 6. Average Success Score by Number of Genres
genre_count_perf = df.groupby('Genre_Count')['Success_Score'].mean()

axes[2,1].plot(genre_count_perf.index, genre_count_perf.values, marker='o', color='#dd8452', linewidth=2)
axes[2,1].set_title('Success Score vs. Number of Genres')
axes[2,1].set_xlabel('Number of Genres per Movie')
axes[2,1].set_ylabel('Average Success Score')
axes[2,1].set_xticks(genre_count_perf.index)
axes[2,1].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('dashboard.png', dpi=200, bbox_inches='tight')
plt.show()

```
